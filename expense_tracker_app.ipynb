#                                              ======EXPENSE TRACKER PROJECT=====

```python
expenses = []

def add_expense():
    category = input("Enter category: ")
    amount = float(input("Enter amount: "))

    if amount > 0:
        expense = {
            "category": category,
            "amount": amount
        }
        expenses.append(expense)
        print("Expense added successfully!")
    else:
        print("Amount must be greater than 0")


def view_expenses():
    if len(expenses) == 0:
        print("No expenses found")
    else:
        for e in expenses:
            print(e["category"], "-", e["amount"])


def total_expense():
    total = 0

    for e in expenses:
        total = total + e["amount"]

    print("Total Expense:", total)


def check_budget():
    budget = float(input("Enter your budget: "))
    total = 0

    for e in expenses:
        total = total + e["amount"]

    remaining = budget - total

    if remaining > 0:
        print("Remaining Budget:", remaining)
    elif remaining == 0:
        print("Your budget is fully used")
    else:
        print("Budget exceeded by:", abs(remaining))


def main():
    while True:
        print("\n===== EXPENSE TRACKER =====")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Total Expense")
        print("4. Check Budget")
        print("5. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            add_expense()
        elif choice == 2:
            view_expenses()
        elif choice == 3:
            total_expense()
        elif choice == 4:
            check_budget()
        elif choice == 5:
            print("======Thank you======")
            break
        else:
            print("Invalid choice")


main()
```
